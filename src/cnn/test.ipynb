```python
import cv2
import numpy as np
from src.models.cnn import CNN
import os
import sys
from pathlib import Path

import matplotlib.pyplot as plt
from dotenv import load_dotenv

cnn = CNN()

ROOT = Path.cwd().parents[1]
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
    

img1_path = ROOT / "public/testframes/lion1.jpg"
img2_path = ROOT / "public/testframes/lion2.jpg"

img1 = cv2.imread(str(img1_path))
img2 = cv2.imread(str(img2_path))

vec1 = cnn.extract(img1)
vec2 = cnn.extract(img2)

cosine = np.dot(vec1, vec2)   # đã normalize nên không cần chia
print(f"Cosine similarity: {cosine:.4f}")
# Kỳ vọng: ~0.10–0.25 (anime girl vs sư tử rất khác nhau về ngữ nghĩa)
```
